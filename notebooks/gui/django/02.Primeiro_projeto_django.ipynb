## Instalação do Django
---

Django é uma biblioteca em Python, e seu processo de instalação segue as mesmas regras de qualquer outra biblioteca Python, ou seja, precisamos de usar o gerenciador de pacotes `pip` em um ambiente virtual `.venv`:

1. Crie uma pasta onde você irá salvar todos os seus projetos feitos com Django.
2. Dentro dessa pasta, crie um ambiente virtual `.venv` digitando o comando `py -m venv .venv` no console.
3. Depois, ative o ambiente virtual com o comando `.venv\Scripts\activate` no console.
4. Coml a venv ativada, instale o Django com o comando `pip install django` no console.
5. Com o Django instalado, agora iremos criar o nosso primeiro projeto...

## Criando uma nova aplicação Django
---

Embora você já tenha criado um projeto Django, você ainda não criou uma aplicação, e é isso o que faremos agora:

1. Abra o console, e dentro dele navegue para dentro da pasta do seu projeto django, usando o comando `cd nome_do_projeto`.
2. Certifique-se de que o seu terminal esteja dentro da pasta do seu projeto, e então execute o comando `py manage.py startapp nome_da_aplicacao`. Isso fará com que uma outra pasta dentro do seu projeto seja criada, com uma outra estrutura. Veja:

## Hello World em Django
---

Agora o nosso próximo passo é fazer um pequeno **Hello World** para ver se conseguiremos visualizar nossas páginas HTML no nosso projeto. Como dissemos anteriormente, o Django trabalha com MTV, ou seja, Model-Template-View, que por sua vez é muito parecido com o famoso MVC (Model-View-Controller). Em ambos os casos a view é o que chamamos de Front-End.

Normalmente, a camada da View é estruturada através de duas pastas: a `template`, que guarda todos os arquivos HTML, e somente os arquivos HTML da sua aplicação, e a pasta `static`, que guarda os outros arquivos que são dependentes do HTML, que são: imagens em geral, áudio e vídeos, CSS e JavaScript. Essas duas pastas são separadas, porém, no mesmo diretório.

1. Abra a pasta da sua aplicação.
2. Dentro da pasta da sua aplicação, crie uma outra pasta chamada `templates` (o nome tem que ser exato).
3. Dentro dessa pasta, crie um arquivo html chamado `index.html`.


<div style="display: flex; justify-content: center">
    <img src="../../assets/django-gif-02-07.gif" alt="Django" />
</div>

4. Abrao arquivo `index.html` e faça um código-fonte básico. Segue abaixo o exemplo que poderemos utilizar:



~~~html
<!DOCTYPE html>
<html lang="pt-br">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Document</title>
</head>
<body>
    <h1>Olá, mundo Django😎🐍</h1>
</body>
</html>
~~~


5. Agora, fora da pasta `templates`, mas ainda dentro da pasta da aplicação, existe um arquivo chamado `views.py`. Esse arquivo permite que o django possa reconhecer os arquivos HTML do seu projeto. Abra o arquivo `views.py`.
6. Digite o código-fonte abaixo em `views.py`:

```python
from django.shortcuts import render

# Create your views here.
def index(request):
    return render(request, "index.html")
```

7. Ótimo. Agora, você deve ter percebido que dentro do core do projeto existe um arquivo chamado `urls.py`. Iremos precisar dele mais para frente. O que importa agora é que precisaremos de um outro arquivo com esse mesmo nome dentro da pasta da aplicação, e esse arquivo ainda não existe.
8. Portanto, dentro da pasta da aplicação, crie um novo arquivo chamado `urls.py`.

<div style="display: flex; justify-content: center">
    <img src="../../assets/django-gif-02-08.gif" alt="Django" />
</div>

9. No arquivo `urls.py`, faça o código-fonte abaixo:

```python
from django.urls import path
from . import views

urlpatterns = [
    path('', views.index, name='index'),
]
```

10. Isso indicará para o Django que o endereço padrão da aplicação será o arquivo `index.html` que criamos.
11. Agora precisaremos indicar para o Django que a nossa aplicação faz parte do ecosistema do projeto. Para isso, vá no diretório do core da aplicação, ou seja, a pasta com o nome do seu projeto, e dentro dela abra o arquivo `settings.py`:

<div style="display: flex; justify-content: center">
    <img src="../../assets/django-02-09.gif.png" alt="Django" />
</div>

12. Esse arquivo guarda a configuração padrão do seu projeto, que todos os apps ligados a ele precisarão seguir. Pois bem, faremos agora algumas alterações nesse arquivo.
13. Desça para a linha onde está escrito `INSTALLED_APPS`. Esse código é de uma lista de apps instalados no seu projeto. Repare que ele ainda não tem a aplicação que estamos desenvolvendo, e portanto precisaremos incluí-la aí. Acrescente na lista o nome do seu app. Exemplo, se o seu app se chaama `primeiro_app`, então a lista em `settings.py` ficará assim:

```python
INSTALLED_APPS = [
    'django.contrib.admin',
    'django.contrib.auth',
    'django.contrib.contenttypes',
    'django.contrib.sessions',
    'django.contrib.messages',
    'django.contrib.staticfiles',
    'primeiro_app' # app acrescentado
]
```

14. Agora só falta adicionarmos o caminho da home page da nossa aplicação no core do Django. Lembra quando eu disse mais cedo que existe dentro do core do projeto um outro arquivo chamado `urls.py` que já vinha no projeto? Pois bem, agora sim iremos precisar dele. Abra-o:


<div style="display: flex; justify-content: center">
    <img src="../../assets/django-02-10.png" alt="Django" />
</div>

15. Dentro desse arquivo haverá outra lista: o `urlpatterns`. Iremos adicionar outro elemento: `path('', include('nome_do_projeto.urls'))`.
16. Não se esqueça também de importar o comando `include` na biblioteca `django.urls`.
17. Supondo que o seu app se chama `primeiro_app`, o código-fonte desse arquivo ficará assim:

```python
from django.contrib import admin
from django.urls import path, include

urlpatterns = [
    path('admin/', admin.site.urls),
    path('', include('primeiro_app.urls')),
]
```

18. Agora podemos testar a home page da nossa aplicação. Rode no console o comando `py manage.py runserver` e depois acesse `localhost:8000` no seu navegador, ou atualize a página, caso já esteja aberta.
