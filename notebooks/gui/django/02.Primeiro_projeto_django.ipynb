## Instalação do Django
---

Django é uma biblioteca em Python, e seu processo de instalação segue as mesmas regras de qualquer outra biblioteca Python, ou seja, precisamos de usar o gerenciador de pacotes `pip` em um ambiente virtual `.venv`:

1. Crie uma pasta onde você irá salvar todos os seus projetos feitos com Django.
2. Dentro dessa pasta, crie um ambiente virtual `.venv` digitando o comando `py -m venv .venv` no console.
3. Depois, ative o ambiente virtual com o comando `.venv\Scripts\activate` no console.
4. Coml a venv ativada, instale o Django com o comando `pip install django` no console.
5. Com o Django instalado, agora iremos criar o nosso primeiro projeto...

## Hello World em Django
---

Agora o nosso próximo passo é fazer um pequeno **Hello World** para ver se conseguiremos visualizar nossas páginas HTML no nosso projeto. Como dissemos anteriormente, o Django trabalha com MTV, ou seja, Model-Template-View, que por sua vez é muito parecido com o famoso MVC (Model-View-Controller). Em ambos os casos a view é o que chamamos de Front-End.

Normalmente, a camada da View é estruturada através de duas pastas: a `template`, que guarda todos os arquivos HTML, e somente os arquivos HTML da sua aplicação, e a pasta `static`, que guarda os outros arquivos que são dependentes do HTML, que são: imagens em geral, áudio e vídeos, CSS e JavaScript. Essas duas pastas são separadas, porém, no mesmo diretório.

1. Abra a pasta da sua aplicação.
2. Dentro da pasta da sua aplicação, crie uma outra pasta chamada `templates` (o nome tem que ser exato).
3. Dentro dessa pasta, crie um arquivo html chamado `index.html`. Faça um código-fonte básico. Segue abaixo o exemplo que poderemos utilizar:

~~~html
<!DOCTYPE html>
<html lang="pt-br">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Document</title>
</head>
<body>
    <h1>Olá, mundo Django😎🐍</h1>
</body>
</html>
~~~
4. Agora, fora da pasta `templates`, mas ainda dentro da pasta da aplicação, existe um arquivo chamado `views.py`. Esse arquivo permite que o django possa reconhecer os arquivos HTML do seu projeto. Abra o arquivo `views.py`.
5. Digite o código-fonte abaixo em `views.py`:

```python
from django.shortcuts import render

# Create your views here.
def index(request):
    return render(request, "index.html")
```
