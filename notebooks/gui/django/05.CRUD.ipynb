# **CRUD**
---

Neste aula iremos aprender a criar um CRUD. Mas o que é um CRUD?

> CRUD é o tipo de sistema computacional mais básico que existe envolvendo banco de dados. É uma sigla que significa:
> - **C**reate
> - **R**ead
> - **U**pdate
> - **D**elete
> 
> Em outras palavras: Cadastrar, Listar, Alterar e Excluir.

A página de admin já consegue fazer tudo isso, mas como dito anteriormente, precisaremos dispor esses dados para o usuário final. Vamos começar pela nossa home page, que irá listar todos os usuários cadastrados no nosso sistema.

## Read
---

Normalmente começaríamos pelo **CREATE**, mas como já temos alguns dados cadastrados (e já temos como cadastrar mais pelo admin), vamos fazer primeiro o **READ**.

1. Abra o arquivo `index.html`.
2. No momento, o arquivo deverá estar com o seguinte código-fonte:

~~~html
{% extends 'base.html' %}
{% block title %}Página Inicial{% endblock %}
{% block content %}
    <main>
        <h1>Olá, mundo Django😎🐍</h1>
    </main>
{% endblock %}
~~~

3. Altere o código-fonte dentro de `<main>` para o que está logo abaixo:

~~~html
<div class="container">
    <h2>Lista de Usuários</h2>
    <ul>
        {% for usuario in pessoas %}
            <li><b>ID:</b>{{ usuario.id_pessoa }} | <b>Nome:</b>{{ usuario.nome }} | <b>E-mail:</b>{{ usuario.email }} | <b>CPF:</b>{{ usuario.cpf }} | <b>Peso:</b>{{ usuario.peso }} kg | <b>Altura:</b>{{ usuario.altura }} m | <b>Data de nascimento:</b>{{ usuario.data_nascimento }}</li>
        {% empty %}
            <li>Nenhum usuário cadastrado.</li>
        {% endfor %}
    </ul>
</div>
~~~

4. Esse código é o responsável por listar todos os usuários cadastrados no banco de dados, mas ele ainda não fará isso, pois ele ainda não reconhece o nome `usuario` como objeto Python.
5. Para isso, precisaremos alterar o render em `views.py`.
6. Abra `views.py` e vá na função `index`.
7. Altere a função para o código abaixo:

```python
def index(request):
    usuarios = Usuario.objects.all()
    return render(request, "index.html", {"usuarios": usuarios})
```

8. Não se esqueça de importar a classe Usuario:

```python
from .models import Usuario
```

9. O código-fonte desse arquivo até o momento ficará assim:

```python
from django.shortcuts import render
from .models import Usuario

# Create your views here.
def index(request):
    usuarios = Usuario.objects.all()
    return render(request, "index.html", {"usuarios": usuarios})

def cadastrar_usuario(request):
    return render(request, 'cadastrar.html')
```
