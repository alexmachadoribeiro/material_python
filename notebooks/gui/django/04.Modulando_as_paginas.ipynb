## Montando a estrutura da templates
---

Vimos nas aulas anteriores que os arquivos HTML ficam sempre dentro da pasta `templates`, que por sua vez fica no diretório da aplicação do sistema. Agora vamos estruturar essa pasta para organizar o nosso Front-End.

Lembre-se de que nessa pasta só podemos guardar os arquivos HTML, nada mais, e nem os arquivo HTML podem ser salvos em outra pasta que não seja a templates.

6. Volte no arquivo `base.html` e cole o código.
7. O código-fonte do arquivo `base.html` nesse momento deverá estar assim:

~~~html
<!doctype html>
<html lang="en">
  <head>
    <meta charset="utf-8">
    <meta name="viewport" content="width=device-width, initial-scale=1">
    <title>Bootstrap demo</title>
    <link href="https://cdn.jsdelivr.net/npm/bootstrap@5.3.8/dist/css/bootstrap.min.css" rel="stylesheet" integrity="sha384-sRIl4kxILFvY47J16cr9ZwB07vP4J8+LH7qKQnuqkuIAvNWLzeN8tE5YBujZqJLB" crossorigin="anonymous">
  </head>
  <body>
    <h1>Hello, world!</h1>
    <script src="https://cdn.jsdelivr.net/npm/bootstrap@5.3.8/dist/js/bootstrap.bundle.min.js" integrity="sha384-FKyoEForCGlyvwx9Hj09JcYn3nv7wiPVlz7YYwJrWVcXK/BmnVDxM+D2scQbITxI" crossorigin="anonymous"></script>
  </body>
</html>
~~~

8. Agora vamos fazer algumas alterações nesse código. Na linha 2, onde está `<html lang="en">`, altere o idioma em `lang` para `lang=pt-br`, ficando `<html lang="pt-br">`.
9. Para conseguirmos usar código Python dentro do HTML, precisamos do **Jinja**. O Jinja é um recurso do Python que já vem incluso no Django, e não pede instalação extra. O que ele faz é incluir comandos Python dentro do HTML. Sua estrutura é `{% comando %}`.
10. Por exemplo, vamos pedir para que o título da página (tag `<title>`) mude toda vez que o conteúdo mudar. Para isso precisaremos inserir um comando Jinja dentro da tag `<title>`. Na linha 6, onde tem o comando `<title>Bootstrap demo</title>`, dentro de `<title>`, insira o comando `{% block title %}Meu projeto Django{% endblock %}`. O comando ficará assim: `<title>{% block title %}Meu projeto Django{% endblock %}</title>`.
11. Dentro de `<body>`, iremos dividir a nossa página em 3 partes:
- header (onde ficará o menu superior)
- main (onde ficará o conteúdo principal da página)
- footer (onde ficará o rodapé da página)
12. Para isso, iremos apagar o código `<h1>Hello, World!</h1>` e inserir o seguinte código dentro de `<body>`:

~~~html
{% include 'header.html' %}
{% block content %}{% endblock %}
{% include 'footer.html' %}
~~~

13. Ao final, o código-fonte do arquivo `base.html` deverá ser esse:

~~~html
<!doctype html>
<html lang="pt-br">
    <head>
        <meta charset="utf-8">
        <meta name="viewport" content="width=device-width, initial-scale=1">
        <title>{% block title %}Meu projeto Django{% endblock %}</title>
        <link href="https://cdn.jsdelivr.net/npm/bootstrap@5.3.8/dist/css/bootstrap.min.css" rel="stylesheet" integrity="sha384-sRIl4kxILFvY47J16cr9ZwB07vP4J8+LH7qKQnuqkuIAvNWLzeN8tE5YBujZqJLB" crossorigin="anonymous">
    </head>

    <body>
        {% include 'header.html' %}
        {% block content %}{% endblock %}
        {% include 'footer.html' %}
        <script src="https://cdn.jsdelivr.net/npm/bootstrap@5.3.8/dist/js/bootstrap.bundle.min.js" integrity="sha384-FKyoEForCGlyvwx9Hj09JcYn3nv7wiPVlz7YYwJrWVcXK/BmnVDxM+D2scQbITxI" crossorigin="anonymous"></script>
    </body>
</html>
~~~

21. Não vamos testar ele por hora. Agora volte ao arquivo `index.html` que havíamos criado em outra aula, pois agora iremos modificá-lo.

{% endblock %}
~~~
