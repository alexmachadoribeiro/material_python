# **Trabalhando com o Front-End do sistema**
---

Nessa aula, iremos trabalhar diretamente com o Front-End da aplicação. Vamos aprender como fazer com que apenas uma parte do conteúdo da página seja carregado pela aplicação, fazendo com que as páginas abram muito mais rápido para o usuário final.

## Montando a estrutura da templates
---

Vimos nas aulas anteriores que os arquivos HTML ficam sempre dentro da pasta `templates`, que por sua vez fica no diretório da aplicação do sistema. Agora vamos estruturar essa pasta para organizar o nosso Front-End.

1. A pasta `templates` já existe. Inclusive, dentro dela já colocamos um arquivo chamado `index.html` que nesse momento está servindo de home page para a nossa aplicação. Lembre-se de que nessa pasta só podemos guardar os arquivos HTML, nada mais, e nem os arquivo HTML podem ser salvos em outra pasta que não seja a templates.
2. Abra a pasta `templates`, e dentro dela crie um novo arquivo HTML chamado `base.html`. O objetivo desse arquivo é servir de esqueleto para todos os arquivos HTML que teremos na nossa apliacção.

4. No site do Bootstrap, na barra de menu superior da página, clique em **Docs**:

6. Volte no arquivo `base.html` e cole o código.
7. O código-fonte do arquivo `base.html` nesse momento deverá estar assim:

~~~html
<!doctype html>
<html lang="en">
  <head>
    <meta charset="utf-8">
    <meta name="viewport" content="width=device-width, initial-scale=1">
    <title>Bootstrap demo</title>
    <link href="https://cdn.jsdelivr.net/npm/bootstrap@5.3.8/dist/css/bootstrap.min.css" rel="stylesheet" integrity="sha384-sRIl4kxILFvY47J16cr9ZwB07vP4J8+LH7qKQnuqkuIAvNWLzeN8tE5YBujZqJLB" crossorigin="anonymous">
  </head>
  <body>
    <h1>Hello, world!</h1>
    <script src="https://cdn.jsdelivr.net/npm/bootstrap@5.3.8/dist/js/bootstrap.bundle.min.js" integrity="sha384-FKyoEForCGlyvwx9Hj09JcYn3nv7wiPVlz7YYwJrWVcXK/BmnVDxM+D2scQbITxI" crossorigin="anonymous"></script>
  </body>
</html>
~~~

8. Agora vamos fazer algumas alterações nesse código. Na linha 2, onde está `<html lang="en">`, altere o idioma em `lang` para `lang=pt-br`, ficando `<html lang="pt-br">`.
9. Para conseguirmos usar código Python dentro do HTML, precisamos do **Jinja**. O Jinja é um recurso do Python que já vem incluso no Django, e não pede instalação extra. O que ele faz é incluir comandos Python dentro do HTML. Sua estrutura é `{% comando %}`.
10. Por exemplo, vamos pedir para que o título da página (tag `<title>`) mude toda vez que o conteúdo mudar. Para isso precisaremos inserir um comando Jinja dentro da tag `<title>`. Na linha 6, onde tem o comando `<title>Bootstrap demo</title>`, dentro de `<title>`, insira o comando `{% block title %}Meu projeto Django{% endblock %}`. O comando ficará assim: `<title>{% block title %}Meu projeto Django{% endblock %}</title>`.
11. Dentro de `<body>`, iremos dividir a nossa página em 3 partes:
- header (onde ficará o menu superior)
- main (onde ficará o conteúdo principal da página)
- footer (onde ficará o rodapé da página)
12. Para isso, iremos apagar o código `<h1>Hello, World!</h1>` e inserir o seguinte código dentro de `<body>`:

~~~html
{% include 'header.html' %}
{% block content %}{% endblock %}
{% include 'footer.html' %}
~~~

13. Ao final, o código-fonte do arquivo `base.html` deverá ser esse:

~~~html
<!doctype html>
<html lang="pt-br">
    <head>
        <meta charset="utf-8">
        <meta name="viewport" content="width=device-width, initial-scale=1">
        <title>{% block title %}Meu projeto Django{% endblock %}</title>
        <link href="https://cdn.jsdelivr.net/npm/bootstrap@5.3.8/dist/css/bootstrap.min.css" rel="stylesheet" integrity="sha384-sRIl4kxILFvY47J16cr9ZwB07vP4J8+LH7qKQnuqkuIAvNWLzeN8tE5YBujZqJLB" crossorigin="anonymous">
    </head>

    <body>
        {% include 'header.html' %}
        {% block content %}{% endblock %}
        {% include 'footer.html' %}
        <script src="https://cdn.jsdelivr.net/npm/bootstrap@5.3.8/dist/js/bootstrap.bundle.min.js" integrity="sha384-FKyoEForCGlyvwx9Hj09JcYn3nv7wiPVlz7YYwJrWVcXK/BmnVDxM+D2scQbITxI" crossorigin="anonymous"></script>
    </body>
</html>
~~~

15. Na pasta `templates`, crie um novo arquivo HTML chamado `header.html`, e depois abra-o.
16. Não há a necessidade de criar o esqueleto HTML padrão nesse arquivo. Ele já vai usar o que está dentro do arquivo `base.html` como base.

## Inserindo mídia, css e js no sistema
---

Como dito anteriormente, a pasta templates é usada para os arquivos HTML, mas não podemos utilizá-la para outros tipos de arquivos que normalmente usamos em Front-End. Para isso, teremos que criar outra pasta: a `static`. Essa pasta deve ser separada da `templates`, porém, dentro do mesmo diretório. É dentro da `static` que iremos colocar as pastas `img`, `css`, e `js`, por exemplo.

1. Dentro da pasta da aplicação, crie uma nova pasta chamada exatamente de `static`:

<div style="display: flex; justify-content: center">

## Mudando o conteúdo principal da página
---

Agora, vamos trabalhar com mudança de página. Como o Django faz para mudar o conteúdo de uma página para outra? Veremos isso a seguir.

1. Lembra que estamos fazendo um sistema de cadastro de usuários para o nosso cliente? Isso significa que devemos ter uma página de cadastro. Aquela que vimos na seção do admin não vale, pois aquela só será útil para o administrador do sistema, que não é o usuário final.
2. Dentro da pasta `templates`, crie um novo arquivo chamado `cadastro.html`.
3. Escreva o código abaixo em `cadastro.html`:

~~~html
{% extends 'base.html' %}
{% block title %}Cadastro de Pessoas{% endblock %}
{% block content %}
    <h2>Cadastro de Pessoas</h2>
    <form method="POST" action="">
        {% csrf_token %}
        <label for="nome">Nome:</label>
        <input type="text" id="nome" name="nome" required>
        <br><br>
        <label for="email">Email:</label>
        <input type="email" id="email" name="email" required>
        <br><br>
        <label for="cpf">CPF:</label>
        <input type="text" id="cpf" name="cpf" required>
        <br><br>
        <label for="peso">Peso:</label>
        <input type="number" step="0.01" id="peso" name="peso" placeholder="kg" required>
        <br><br>
        <label for="altura">Altura:</label>
        <input type="number" step="0.01" id="altura" name="altura" placeholder="m" required>
        <br><br>
        <label for="data_nascimento">Data de Nascimento:</label>
        <input type="date" id="data_nascimento" name="data_nascimento" required>
        <br><br>
        <button type="submit">Cadastrar</button>
    </form>
{% endblock %}
~~~

4. Mude o código-fonte conforme o gosto, fazendo uso dos benefícios do Bootstrap.
5. Caso seja necessário, acrescente `{% load static %}` no início do código também.
6. Há muita coisa para preparar nesse formulário, mas faremos isso nas próximas aulas. Por hora, só queremos que ele seja exibido. Portanto, abra o arquivo `views.py`.
7. Adicione a função abaixo em `views.py`:

```python
def cadastrar_usuario(request):
    return render(request, 'cadastrar.html')
```

8. O código-fonte de `views.py` nesse momento ficará assim:

```python
from django.shortcuts import render

# Create your views here.
def index(request):
    return render(request, "index.html")

def cadastrar_usuario(request):
    return render(request, 'cadastrar.html')
```

9. Agora, abra o arquivo `urls.py` que fica dentro da pasta da aplicação (lembre-se, aquela que foi criada por nós, e não a que já veio com o projeto).

11. O código-fonte do arquivo ficará assim:

```python
from django.urls import path
from . import views

urlpatterns = [
    path('', views.index, name='index'),
    path('cadastrar/', views.cadastrar_usuario, name='cadastrar_usuario'),
]
```
