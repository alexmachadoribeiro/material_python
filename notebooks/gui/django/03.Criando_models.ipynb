## Criando e configurando novo banco de dados
---

A primeira coisa a se fazer é decidir qual SGBD (Sistema Gerenciador de Banco de Dados) iremos utilizar. Para fins didáticos, iremos utilizar o **SQLite**. O SQLite é um SGBD Relacional de pequeno porte que pode ser utilizado para fins de testes ou para sistemas pequenos. Ele cria um arquivo onde as tabelas e os dados serão armazenados, não havendo a necessidade de configurar e/ou se conectar a um servidor na rede, já que ele pode rodar dentro da própria aplicação.

Os principais motivos pelos quais vamos usar o SQLite é a pouca exigência de configuração, o fato de rodar localmente, e também por ser o SGBD padrão que já vem instalado junto com o Python, diferente de outros SGBDs em que teríamos que instalar outros programas. Vamos começar?

1. No seu projeto Django, dentro do core do projeto, abra a pasta `settings.py`:

<div style="display: flex; justify-content: center">
    <img src="../../assets/django-02-09.gif.png" alt="Django" />
</div>

2. Desça a barra de rolagem até encontrar o dicionário `DATABASES` (possívelmente na linha 76). Se encontrar, ele deve se parecer com o código abaixo:

```python
DATABASES = {
    'default': {
        'ENGINE': 'django.db.backends.sqlite3',
        'NAME': BASE_DIR / 'db.sqlite3',
    }
}
```

3. Essa configuração já é o padrão do próprio Django. Ele automaticamente irá se conectar ao SQLite que já vem com o Python nativo do sistema, e irá criar um database na raíz do seu projeto chamado `db.sqlite3`. Isso se você não tiver feito absolutamente na da em relação ao banco de dados no seu projeto.
4. Inclusive, caso já tenha executado o seu projeto anteriormente (e se estiver seguindo as minhas aulas, seu projeto já foi executado pelo menos 2 vezes), ele já vai ter gerado automaticamente esse banco. Você pode reparar no explorador de arquivos que na raíz do projeto tem um arquivo chamado `db.sqlite3`:

<div style="display: flex; justify-content: center">
    <img src="../../assets/django-02-12.png" alt="Django" />
</div>

5. Acontece que esse não é o banco que queremos usar no nosso projeto, então você poderá deletá-lo sem maiores problemas, pois iremos configurar o nosso próprio banco.
6. Voltando ao `settings.py`, dentro do dicionário `DATABASE`, na linha em que se encontra a chave `NAME`, altere o nome do banco de `db.sqlite3` para `meu_banco.db`. Se fizer da forma correta, o dicionário deverá ficar assim:

```python
DATABASES = {
    'default': {
        'ENGINE': 'django.db.backends.sqlite3',
        'NAME': BASE_DIR / 'meu_banco.db', # linha a ser alterada
    }
}
```

7. Isso fará com que, ao rodar na próxima execução, ele crie um banco de dados na raíz do projeto chamado `meu_banco.db` ao invés de `db.sqlite3`, como estava anteriormente.
8. Não se preocupe em guardar o seu banco em uma pasta própria para ele, pois embora seja possível, não é recomendado pela própria convenção do Django, sendo o mais recomendado que ele esteja dentro da própria raíz do projeto. Portanto, sigamos em frente.
9. Rode novamente o seu projeto, executando o comando `py manage.py runserver` no terminal.
10. Não precisa verificar sua aplicação no navegador. Já sabemos como o Front-End do projeto está nesse momento, e as alterações que fizemos não irá mudar o visual da página. A única mudança que iremos verificar agora está na estrutura de arquivos e pastas do seu projeto. Veja que apareceu na raíz do seu projeto um novo arquivo, o `meu_banco.db`:

<div style="display: flex; justify-content: center">
    <img src="../../assets/django-02-13.png" alt="Django" />
</div>

11. Essa é a única alteração feita no momento. Inclusive, esse banco está completamente vazio, sem tabela, sem dados. Encerre o servidor com `Ctrl+C`.

## Criando admin do sistema Django
---

Antes de criarmos as nossas entidade do banco do projeto, você já se perguntou quem vai gerenciar esse banco? Você desenvolvedor com certeza não terá tempo para ficar dando manutenção no banco do cliente 24 horas por dia, afinal, você tem uma vida. O usuário, por outro lado, não tem o mesmo conhecimento que você sobre programação, nem sobre SQL para poder gerenciar o sistema dele. Como fazer então?

Se estivéssemos trabalhando com outros frameworks concorrentes, teríamos que criar um sistema de gerenciamento, com um usuário `admin` só para que o cliente possa gerenciar o banco usando uma Interface Gráfica amigável.

A nossa sorte é que estamos usando Django, e o Django, como dito anteriormente, já tem tudo pronto. Ele já vem com um sistema de administrador já pronto, com Front-End e tudo. Só precisamos configurar ele. É o que faremos nessa seção:

1. Abra o terminal na pasta raíz do projeto.
2. Digite o seguinte comando: `py manage.py makemigrations`, e em seguida, digite o comando `py manage.py migrate`, ambos no console.
3. Acostume-se com esses dois comandos: toda vez que fizermos alguma alteração no nosso banco de dados, antes de rodarmos a aplicação, precisaremos rodar esses dois comandos nessa ordem para atualizar o banco do projeto.
4. O próximo passo é criarmos o que chamamos de **Superusuário**. Ele será o admin do sistema no lugar do programador. Assim, ele mesmo poderá fazer alterações no banco sem precisar de nós, e sem que isso seja feita no nível do usuário, aumentando a segurança e a disponibilidade do sistema.
5. Digite no terminal o comando `py manage.py createsuperuser`.
6. Assim que der `Enter`, ele irá pedir para inserir o nome do super usuário. Escolha um nome bem seguro. Para fins didáticos, nesse tutorial iremos escolher o nome `admin` para o super usuário. Portanto, ao ser solicitado, digite `admin`.
7. Assim que definir o nome do super usuário, ele irá pedir para que você informe o e-mail do super usuário. Claro que você irá informar o e-mail de quem for gerenciar o programa, mas por outro lado, ele não irá fazer a verificação desse e-mail, o que significa que, para fins didáticos, poderemos inventar um e-mail que não existe. Vamos usar para o nosso sistema o e-mail `admin@admin.com`. Para o nosso caso, irá funcionar.
8. O próximo passo é inserir a senha. Caso você informe uma senha que não alterne entre letras maiúsculas e minúsculas, não possua números, e não tenha mais do que 8 caracteres, ele irá exibir uma mensagem de erro, mas não se preocupe: podemos usar uma senha simples nesse caso para fins didáticos. Insira (ou não, você que sabe) a senha `admin` e aperte `Enter`.
9. Repita a mesma senha e dê `Enter`.
10. Confirme teclando `y`.
11. Veja o passo a passo na imagem abaixo:

12. Pronto! Super usuário criado com sucesso!!! Agora iremos verificar se existe um sistema de admin no projeto.
13. Execute o comando `py manage.py runserver` para rodar o servidor.
14. Abra o navegador de sua preferência e acesse o endereço `localhost:8000/admin`.
15. Se der certo, irá aparecer a página abaixo:

## Criando as models do sistema
---

Agora serão criadas as entidades do nosso banco. Suponha que estamos desenvolvendo um sistema de cadastro de usuário para um cliente. Logo, precisaremos ter uma entidade no nosso banco chamado `Usuario`. Então, mãos à obra:

1. Dentro da pasta da aplicação (no nosso sistema será a pasta `primeiro_app`), encontre e abra o arquivo `models.py`.
2. No momento ele se encontra com o seguinte código-fonte:

```python
from django.db import models

# Create your models here.
```

3. Suponha que o usuário deverá cadastrar suas informações. As informações exigidas serão essas:
- Nome
- E-mail
- CPF
- Peso
- Altura
- Data de Nascimento
4. Para isso, precisaremos criar uma classe chamada `Usuario`. Essa classe deverá ter, além desses atributos acima, um outro atributo que irá servir de chave primária para a tabela no banco de dados, do qual chamaremos de `id_usuario`. Logo, iremos acrescentar dentro de `models.py` o código-fonte abaixo:

```python
class Usuario(models.Model):
    id_usuario = models.AutoField(primary_key=True)
    nome = models.CharField(null=False, blank=False)
    email = models.EmailField(unique=True, null=False, blank=False)
    cpf = models.CharField(max_length=14, unique=True, null=False, blank=False)
    peso = models.DecimalField(max_digits=5, decimal_places=2, null=False, blank=False)
    altura = models.DecimalField(max_digits=4, decimal_places=2, null=False, blank=False)
    data_nascimento = models.DateField(null=False, blank=False)

    def __str__(self):
        return self.nome
```

5. Ok. Ainda dentro da pasta da aplicação, existe um outro arquivo chamado `admin.py`. Abra-o. O código-fonte dele é exatamente igual ao código-abaixo:

```python
from django.contrib import admin

# Register your models here.
```

6. Precisaremos registrar a classe `Usuario` nesse arquivo para que ele consiga inserir a tabela no banco de dados. Para isso, acrescente nesse arquivo o código-fonte abaixo:

```python
from .models import Usuario

# Register your models here.
admin.site.register(Usuario)
```

7. Agora execute os comandos `py manage.py makemigrations`, e em seguida, `py manage.py migrate`.

## Relações entre tabelas
---

Quem mexe com banco de dados, sabe que o campo telefone não deve ficar na mesma tabela do usuário por conta da normalização do banco, e devemos aplicar a 1ª FN. Para isso, devemos criar uma nova tabela só para telefone e relacioná-la com o usuário. É o que faremos.

1. Abra novamente o arquivo `models.py`, onde a classe `Usuario` foi criada.
2. Acrescente a classe `Telefone`. Use o código-fonte abaixo:

```python
class Telefone(models.Model):
    id_telefone = models.AutoField(primary_key=True)
    usuario = models.ForeignKey(Usuario, on_delete=models.CASCADE)
    numero = models.CharField(max_length=15, null=False, blank=False)

    def __str__(self):
        return self.numero
```

3. Veja que o campo `usuario` na tabela `Telefone` está como `ForeignKey`, pois ele faz referência à tabela `Usuario`.
4. Nesse momento, o arquivo `models.py` deverá estar com o seguinte código-fonte:

```python
from django.db import models

# Create your models here.
class Usuario(models.Model):
    id_usuario = models.AutoField(primary_key=True)
    nome = models.CharField(null=False, blank=False)
    email = models.EmailField(unique=True, null=False, blank=False)
    cpf = models.CharField(max_length=14, unique=True, null=False, blank=False)
    peso = models.DecimalField(max_digits=5, decimal_places=2, null=False, blank=False)
    altura = models.DecimalField(max_digits=4, decimal_places=2, null=False, blank=False)
    data_nascimento = models.DateField(null=False, blank=False)

    def __str__(self):
        return self.nome

class Telefone(models.Model):
    id_telefone = models.AutoField(primary_key=True)
    usuario = models.ForeignKey(Usuario, on_delete=models.CASCADE)
    numero = models.CharField(max_length=15, null=False, blank=False)

    def __str__(self):
        return self.numero
```

5. Volte em `admin.py` e registre a nova tabela `Telefone`. O código-fonte de `admin.py` deverá estar assim:

```python
from django.contrib import admin
from .models import Usuario, Telefone

# Register your models here.
admin.site.register(Usuario)
admin.site.register(Telefone)
```

6. Execute `py manage.py makemigrations`, e depois `py manage.py migrate`, para que a nova tabela seja incluída no banco de dados.
7. Rode novamente o servidor com `py manage.py runserver` e volte em `localhost:8000/admin`.
